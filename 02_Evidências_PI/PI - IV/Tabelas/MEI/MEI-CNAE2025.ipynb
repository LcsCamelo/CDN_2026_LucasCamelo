```python
import pandas as pd
import requests

# 1. Carregar o arquivo
df = pd.read_csv(r'C:/Users/Usuario/Documents/Fatec/PI 4/MEI/Dados MEI/MEIVotCNAE.csv', sep=",",encoding='latin-1')



# Renomeia a coluna que contém a informação composta e a coluna de total
df.columns = ['Munícipio_MEI_Código_Total', 'Total','CNAE']

# 3. Separar a primeira coluna pelo hífen ('-')
# Isso separa 'VOTORANTIM' do restante (ex: '0121101,2')
df[['Cidade', 'CNAE']] = df['Munícipio_MEI_Código_Total'].str.split('-', expand=True)
df
# 4. Separar o Código MEI do Total (se for o caso)
# Baseado na sua imagem, o 'Total' foi separado no passo 2
# Vamos separar o 'Código_MEI' do 'Total' *que ainda estava na primeira coluna*, se houver.
# A imagem inicial sugere que a coluna 1 pode conter o total, mas a primeira coluna
# tem a forma 'VOTORANTIM-0121101,2'. Vamos refinar a separação para garantir:

# 5. Selecionar e exibir as colunas finais
df = df[['Cidade', 'CNAE', 'Total']]

df = df.drop(df.index[-1])

df
```

```python
url = 'https://servicodados.ibge.gov.br/api/v2/cnae/subclasses'#url da Query Builder do IBGE

CNAE = requests.get(url).json()
CNAET = pd.json_normalize(CNAE, sep='_')


CNAET = CNAET.rename(columns={
    'id': 'CNAE',
    'descricao':'Desc CNAE',
    'classe_id':'Cód Classe',
    'classe_descricao':'Desc Classe',
    'classe_grupo_id':'Cód Grupo',
    'classe_grupo_descricao':'Desc Grupo',
    'classe_grupo_divisao_id':'Cód Div',
    'classe_grupo_divisao_descricao':'Desc Div',
    'classe_grupo_divisao_secao_id':'Cód Sec',
    'classe_grupo_divisao_secao_descricao':'Desc Sec'
    })


CNAET = CNAET.drop(columns=['observacoes','atividades','classe_observacoes'])
CNAET.info()
CNAET.head(20)
```

```python
df_merged = pd.merge(df, CNAET, on='CNAE', how='left')

df_merged
```

```python
import numpy as np

mapa_setores = {
    'A': 'Primário',
    'B': 'Industrial',
    'C': 'Industrial',
    'D': 'Serviços',
    'E': 'Serviços',
    'F': 'Serviços',
    'G': 'Serviços',
    'H': 'Serviços',
    'I': 'Serviços',
    'J': 'Serviços',
    'K': 'Serviços',
    'L': 'Serviços',
    'M': 'Serviços',
    'N': 'Serviços',
    'O': 'Administração Pública',
    'P': 'Serviços',
    'Q': 'Serviços',
    'R': 'Serviços',
    'S': 'Serviços',
    'T': 'Serviços',
    'U': 'Serviços'
}

df_Mei = df_merged.copy()

df_Mei['Setor'] = df_Mei['Cód Sec'].map(mapa_setores)

df_Mei['Setor'] = df_Mei['Cód Sec'].map(mapa_setores)

df_Mei['Desc Sec'] = df_Mei['Cód Sec'] + " - " + df_Mei['Desc Sec']

df_Mei['Desc Div'] = np.where(df_Mei['Cód Div'] != 'X', df_Mei['Cód Div'] + " - " + df_Mei['Desc Div'],df_Mei['Desc Div'])

df_Mei['Desc Grupo'] = df_Mei.apply(
    lambda x: f"{x['Cód Grupo'][:-1]}.{x['Cód Grupo'][-1]} {x['Desc Grupo']}" if x['Cód Grupo'] != 'X' else x['Desc Grupo'],
    axis=1
)

df_Mei['Desc Classe'] = df_Mei.apply(
    lambda x: f"{x['Cód Classe'][:2]}.{x['Cód Classe'][2:-1]}-{x['Cód Classe'][-1]} {x['Desc Classe']}",
    axis=1
)

df_Mei['Desc CNAE'] = df_Mei.apply(
    lambda x: f"{x['CNAE'][:2]}.{x['CNAE'][2:4]}-{x['CNAE'][4]}/{x['CNAE'][5:7]} {x['Desc CNAE']}",
    axis=1
)

df_Mei
```
