```python
import pandas as pd
import glob
import os
import re # Módulo para expressões regulares (ótimo para extrair o ano)

# 1. Definir o Caminho e Padrão dos Arquivos
# ATENÇÃO: Substitua este caminho pelo caminho completo da sua pasta 'Dados MEI'
# O caminho visto na imagem é: C:\Users\Usuario\Documents\Fatec\PI 4\MEI\Dados MEI
caminho_base = r'C:\Users\Usuario\Documents\Fatec\PI 4\MEI\Dados MEI'
# Padrão para encontrar todos os arquivos CSV (adapte se tiver outros CSVs na pasta)
padrao_arquivos = os.path.join(caminho_base, 'MEIVot*.csv')

# Lista para armazenar todos os DataFrames processados
lista_dfs = []

# 2. Encontrar os Arquivos e Processar em Loop
# glob.glob() encontra todos os arquivos que correspondem ao padrão
for arquivo_path in glob.glob(padrao_arquivos):
    # A. Extrair o Ano do Nome do Arquivo
    # Exemplo: 'MEIVot2015.csv' -> '2015'
    
    # Pega apenas o nome do arquivo (ex: MEIVot2015.csv)
    nome_arquivo = os.path.basename(arquivo_path)
    
    # Usa expressão regular para encontrar os 4 dígitos do ano
    match_ano = re.search(r'\d{4}', nome_arquivo)
    
    if match_ano:
        ano = match_ano.group(0) # Pega o ano como string (ex: '2015')
        print(f"Processando arquivo: {nome_arquivo} (Ano: {ano})")
        
        try:
            # B. Ler o Arquivo CSV
            # Adapte o encoding (e o 'sep') se for necessário
            df = pd.read_csv(arquivo_path, encoding='utf-8')
            
            # C. Adicionar a Coluna 'Ano'
            df['Ano'] = ano
            
            # D. Adicionar o DataFrame processado à lista
            lista_dfs.append(df)
            
        except Exception as e:
            print(f"ERRO ao ler {nome_arquivo}: {e}")
            
    else:
        print(f"AVISO: Não foi possível extrair o ano do arquivo {nome_arquivo}. Ignorando.")


# 3. Consolidar Todos os DataFrames
if lista_dfs:
    # pd.concat junta todos os DFs da lista verticalmente (por linha)
    df_consolidado = pd.concat(lista_dfs, ignore_index=True)
    
    print("\n✅ Consolidação Concluída!")
    print(f"DataFrame final criado com {len(df_consolidado)} registros.")
    
    # 4. Cálculo do Crescimento Percentual (Ano-a-Ano)
    # A função .pct_change() calcula (Valor Atual - Valor Anterior) / Valor Anterior
    df_consolidado['Crescimento_Percentual'] = df_consolidado['TOTAL'].pct_change()

    # O primeiro registro será NaN (não há ano anterior para comparar), vamos preencher com 0
    df_consolidado['Crescimento_Percentual'] = df_consolidado['Crescimento_Percentual'].fillna(0).round(2)

    # 5. Exportar o Novo Arquivo CSV
    caminho_saida = os.path.join(caminho_base, 'MEI_CONSOLIDADO_TOTAL.csv')
    df_consolidado.to_csv(caminho_saida, index=False, encoding='utf-8')
    
    print(f"Arquivo CSV final exportado para: {caminho_saida}")
else:
    print("\n❌ NENHUM ARQUIVO ENCONTRADO OU PROCESSADO. Verifique o caminho e o padrão.")
```
