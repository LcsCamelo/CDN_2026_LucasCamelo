```python
mapa_traducao = {
    'NC': 'Nível Territorial (Código)',
    'NN': 'Nível Territorial',
    'MC': 'Unidade de Medida (Código)',
    'MN': 'Unidade de Medida',
    'V': 'Valor',
    'D1C': 'Município (Código)',
    'D1N': 'Município',
    'D2C': 'Variável (Código)',
    'D2N': 'Variável',
    'D3C': 'Ano (Código)',
    'D3N': 'Ano',
    'D4C': 'Classificação Nacional de Atividades Econômicas (CNAE 2.0) (Código)',
    'D4N': 'Classificação Nacional de Atividades Econômicas (CNAE 2.0)'
}

def substituir_codigos(dados, mapa):
    """
    Percorre uma lista de dicionários e substitui as chaves (códigos curtos)
    pelos nomes longos definidos no mapa de tradução.
    """
    dados_renomeados = []

    for registro in dados:
        novo_registro = {}
        for chave_curta, valor in registro.items():
            # Tenta encontrar o nome longo no mapa de tradução
            chave_longa = mapa.get(chave_curta, chave_curta)

            # Adiciona o valor ao novo registro com o nome longo
            novo_registro[chave_longa] = valor

        dados_renomeados.append(novo_registro)

    return dados_renomeados
```

```python
# Separa a string no primeiro espaço e pega a primeira parte (o código)
Data_IBGE['Cód CNAE'] = Data_IBGE['Classificação Nacional de Atividades Econômicas (CNAE 2.0)'].str.split(' ', n=1, expand=True)[0]
# Limpa a coluna: mantém dígitos E letras, removendo TODO o resto.
Data_IBGE['Cód CNAE'] = Data_IBGE['Cód CNAE'].str.replace(r'[^a-zA-Z^\d]', '', regex=True)

Data_IBGE.rename(columns={'Classificação Nacional de Atividades Econômicas (CNAE 2.0)':'Desc Classe'}, inplace=True)
print(Data_IBGE[['Cód CNAE','Desc Classe']].head())
```

```python
# Categorizando

Data_IBGE['Cód Classe'] = np.where(
    Data_IBGE['Cód CNAE'].str.len() == 5, Data_IBGE['Cód CNAE'], 'X')

Data_IBGE['Cód Grupo'] = np.where(
    Data_IBGE['Cód CNAE'].str.len() > 2, Data_IBGE['Cód CNAE'].str[:3], 'X')

Data_IBGE['Cód Div'] = np.where(
    Data_IBGE['Cód CNAE'].str.len() > 1, Data_IBGE['Cód CNAE'].str[:2], 'X')

Data_IBGE['Cód Sec'] = np.where(
    Data_IBGE['Cód CNAE'].str.isalpha(), Data_IBGE['Cód CNAE'].str[:1], 'X')

Data_IBGE[['Cód CNAE','Cód Classe','Cód Grupo','Cód Div','Cód Sec']]
```

Vamos baixar agora uma tabela auxiliar do IBGE para nos ajudar na categorização do CNAEs obtidos da Tabela 9528

```python
url = 'https://servicodados.ibge.gov.br/api/v2/cnae/grupos'#url da Query Builder do IBGE

CNAE = requests.get(url).json()
CNAET = pd.json_normalize(CNAE, sep='_')
CNAET = CNAET.rename(columns={
    'id': 'Cód Grupo',
    'descricao':'Desc Grupo',
    'divisao_id':'Cód Div',
    'divisao_secao_id':'Cód Sec',
    'divisao_descricao':'Desc Div',
    'divisao_secao_descricao':'Desc Sec'
    })
CNAET = CNAET.drop(columns=['observacoes'])
CNAET.info()
CNAET.head(20)
```

```python

df_base = Data_IBGE.copy()

# Remove duplicatas nas tabelas auxiliares
CNAET_grupo = CNAET[['Cód Grupo', 'Desc Grupo']].drop_duplicates()
CNAET_div = CNAET[['Cód Div', 'Desc Div']].drop_duplicates()
CNAET_divsec = CNAET[['Cód Div', 'Cód Sec']].drop_duplicates()
CNAET_sec = CNAET[['Cód Sec', 'Desc Sec']].drop_duplicates()

# Faz os merges garantindo que não multiplique linhas
df_merged = pd.merge(df_base, CNAET_grupo, on='Cód Grupo', how='left')
assert df_merged.shape[0] == df_base.shape[0], "Linhas multiplicaram no merge do Grupo!"

df_merged = pd.merge(df_merged, CNAET_div, on='Cód Div', how='left')
assert df_merged.shape[0] == df_base.shape[0], "Linhas multiplicaram no merge da Divisão!"

df_merged = pd.merge(df_merged, CNAET_divsec, on='Cód Div', how='left')
condicao_preencher = (df_merged['Cód Sec_x'] == 'X') | (df_merged['Cód Sec_x'].isna())
df_merged.loc[condicao_preencher, 'Cód Sec_x'] = df_merged['Cód Sec_y']
df_merged = df_merged.rename(columns={'Cód Sec_x': 'Cód Sec'})
df_merged = df_merged.drop(columns=['Cód Sec_y'])

df_merged = pd.merge(df_merged, CNAET_sec, on='Cód Sec', how='left')
assert df_merged.shape[0] == df_base.shape[0], "Linhas multiplicaram no merge da Divisão!"

df_merged[['Desc Sec','Desc Div','Desc Grupo']] = df_merged[['Desc Sec','Desc Div','Desc Grupo']].fillna('X')

Data_IBGET = df_merged.copy()

Data_IBGET.head(20)
```

```python
mapa_setores = {
    'A': 'Primário',
    'B': 'Industrial',
    'C': 'Industrial',
    'D': 'Serviços',
    'E': 'Serviços',
    'F': 'Serviços',
    'G': 'Serviços',
    'H': 'Serviços',
    'I': 'Serviços',
    'J': 'Serviços',
    'K': 'Serviços',
    'L': 'Serviços',
    'M': 'Serviços',
    'N': 'Serviços',
    'O': 'Administração Pública',
    'P': 'Serviços',
    'Q': 'Serviços',
    'R': 'Serviços',
    'S': 'Serviços',
    'T': 'Serviços',
    'U': 'Serviços'
}

Data_IBGET['Setor'] = Data_IBGET['Cód Sec'].map(mapa_setores)

Data_IBGET['Desc Sec'] = Data_IBGET['Cód Sec'] + " - " + Data_IBGET['Desc Sec']

Data_IBGET['Desc Div'] = np.where(Data_IBGET['Cód Div'] != 'X', Data_IBGET['Cód Div'] + " - " + Data_IBGET['Desc Div'],Data_IBGET['Desc Div'])


Data_IBGET['Desc Grupo'] = Data_IBGET.apply(
    lambda x: f"{x['Cód Grupo'][:-1]}.{x['Cód Grupo'][-1]} {x['Desc Grupo']}" if x['Cód Grupo'] != 'X' else x['Desc Grupo'],
    axis=1
)

Data_IBGET
```

Vamos criar uma nova métrica relacionando Pessoal ocupado assalariado e Salários e outras remunerações para futuramente calcular o Salário médio por pessoa.

```python
Data_IBGETP['Salário Médio'] = np.where(
    (Data_IBGETP['Pessoal ocupado assalariado'].isnull()) | (Data_IBGETP['Pessoal ocupado assalariado'] == 0 ),
    0,
    (Data_IBGETP['Salários e outras remunerações']/Data_IBGETP['Pessoal ocupado assalariado'])*1000/12
)
df_top_15 = Data_IBGETP.sort_values(by='Salário Médio', ascending=False).head(15)
df_top_15.head(15)
```

```python
Data_IBGETP['Crescimento Anual Pessoal Ocupado Total'] = Data_IBGETP.groupby('Cód CNAE')['Pessoal ocupado total'].pct_change().fillna(0).astype(float)
Data_IBGETP['Crescimento Anual Número de unidades locais'] = Data_IBGETP.groupby('Cód CNAE')['Número de unidades locais'].pct_change().fillna(0).astype(float)
Data_IBGETP['Crescimento Anual Salário Médio'] = Data_IBGETP.groupby('Cód CNAE')['Salário Médio'].pct_change().fillna(0).astype(float)

Data_IBGETP
```
