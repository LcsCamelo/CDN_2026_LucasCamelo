```python
import pandas as pd
import os

# Caminho da pasta onde estão os arquivos Excel
pasta = r"C:\Users\Usuario\Documents\Fatec\PI 4\Turismo"


import pandas as pd
import os

# Caminho da pasta onde estão os arquivos Excel
pasta = r"C:\Users\Usuario\Documents\Fatec\PI 4\Turismo"

# Lista de municípios para filtrar
municipios = ["Araçoiaba da Serra", "Iperó", "Salto de Pirapora", "Sorocaba", "Votorantim"]

# Lista para armazenar os DataFrames
dados_consolidados = []

# Percorrer todos os arquivos Excel na pasta
arquivos = [f for f in os.listdir(pasta) if f.endswith(".xlsx") and not f.startswith("~$")]

print(f"Encontrados {len(arquivos)} arquivos para processar.\n")

for i, arquivo in enumerate(arquivos, start=1):
    caminho_arquivo = os.path.join(pasta, arquivo)
    print(f"[{i}/{len(arquivos)}] Processando: {arquivo} ...")
    
    try:
        # Ler o arquivo Excel
        df = pd.read_excel(caminho_arquivo)
        
        # Verificar se as colunas existem
        colunas_necessarias = ["Atividade Turística", "Número de Inscrição do CNPJ", "Nome Fantasia", "Município"]
        if all(col in df.columns for col in colunas_necessarias):
            # Filtrar pelos municípios desejados
            df_filtrado = df[df["Município"].isin(municipios)]
            
            # Selecionar apenas as colunas necessárias
            df_final = df_filtrado[colunas_necessarias]
            
            # Adicionar coluna com nome do arquivo de origem
            df_final["Arquivo Origem"] = arquivo
            
            # Adicionar à lista
            dados_consolidados.append(df_final)
            
            print(f"   → {len(df_final)} linhas adicionadas.")
        else:
            print("   → Colunas necessárias não encontradas. Pulando arquivo.")
    
    except Exception as e:
        print(f"   → Erro ao processar {arquivo}: {e}")

# Concatenar todos os DataFrames
if dados_consolidados:
    resultado = pd.concat(dados_consolidados, ignore_index=True)
    caminho_saida = os.path.join(pasta, "consolidado.csv")
    resultado.to_csv(caminho_saida, index=False, encoding="utf-8-sig")
    print(f"\n✅ Arquivo consolidado salvo com sucesso em: {caminho_saida}")
else:
    print("\n⚠ Nenhum dado foi consolidado. Verifique as colunas e filtros.")

```
